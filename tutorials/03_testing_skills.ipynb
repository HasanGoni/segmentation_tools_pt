# Tutorial 03 — Testing Your Copilot Skills

This notebook shows how to validate the `SKILL.md` files under `.github/skills/` so you know whether a new skill is discoverable and well-formed before you commit it.

## Why test skills?

A skill can fail silently because:

- The YAML frontmatter has an unescaped colon or a tab indent.
- The `name` field does not match the folder name.
- The `description` is missing or does not contain the trigger words you expect.
- A referenced `./scripts/` or `./assets/` file does not exist.
- The skill is set to neither slash-invokable nor auto-loadable.

The `skill-smoke-test` skill bundles a notebook that catches all of these automatically.

## Run the smoke-test notebook

1. Open `.github/skills/skill-smoke-test/notebooks/test_skills.ipynb`.
2. Run all cells.
3. Read the PASS/FAIL report.

The report also prints suggested prompts like:

```text
"optimize my model for small defects"
```

Copy one, paste it into Copilot Chat, and check whether the right skill loads.

## Add a new skill

1. Create `.github/skills/<my-skill>/SKILL.md`.
2. Re-run `test_skills.ipynb`.
3. Fix any FAIL items.
4. Do a manual chat test with the prompt the notebook suggests.
5. Commit and push.

## Good `description` habits

The `description` field is how Copilot decides to load the skill. Include:

- A one-sentence summary.
- A `Use when...` clause with quoted trigger phrases, e.g. `"my defects are tiny"`.
- Honest status for placeholders: `PLACEHOLDER skill — X is NOT implemented yet.`

## What the smoke test cannot catch

It validates **structure**, not model dispatch. A perfect skill can still fail to load if Copilot decides another skill or the default agent is a better match. Always perform the manual chat test for new skills.
